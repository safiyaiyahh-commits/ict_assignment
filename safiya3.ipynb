```python
def max_of_three(a,b,c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c
num1 = 32
num2 = 45
num3 = 95
print("The maximum number is:", max_of_three(num1,num2,num3))
```
